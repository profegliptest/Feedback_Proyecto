```python
import pandas as pd
import random
from datetime import datetime, timedelta

# 1. Configuración de datos maestros
n_registros = 1000
paises = ['Colombia', 'México', 'Argentina', 'España', 'Chile', 'Perú']
productos_data = {
    'GASEOSA': [('COCACOLA', 1.50), ('PEPSI', 1.25)],
    'ENERGIZANTE': [('SPEED MAX', 2.00), ('REDBULL', 3.50)]
}

data = []

# 2. Bucle para generar los 1000 registros
for _ in range(n_registros):
    # Selección de categoría y producto
    categoria = random.choice(list(productos_data.keys()))
    producto, precio_base = random.choice(productos_data[categoria])

    # Cálculos financieros
    precio_unidad = round(precio_base * random.uniform(0.95, 1.10), 2)
    cantidad = random.randint(1, 100)
    precio_total = round(precio_unidad * cantidad, 2)

    # Fecha aleatoria (últimos 12 meses)
    fecha = datetime.now() - timedelta(days=random.randint(0, 365))

    data.append({
        'Fecha creacion': fecha.strftime('%d/%m/%Y'),
        'Categoria': categoria,
        'Producto': producto,
        'Precio unidad': precio_unidad,
        'cantidad': cantidad,
        'precio total': precio_total,
        'pais': random.choice(paises)
    })

# 3. Crear DataFrame y exportar a Excel
df = pd.DataFrame(data)

# Requerido: pip install openpyxl
nombre_archivo = 'ventas_productos.xlsx'
df.to_excel(nombre_archivo, index=False, engine='openpyxl')

print(f"✅ ¡Archivo '{nombre_archivo}' generado con éxito con {n_registros} registros!")
```
